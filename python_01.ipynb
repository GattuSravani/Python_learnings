```python
l1 = ["sai", 24, "sravani","Balu","Kumari",1,"32"]
l2 = []
l3 = []

for i in l1:
    if type(i) == int:
        l2.append(i)
    else:
        l3.append(i)
print(l2)
l3
```

```python
l4 = [i for i in l1 if type(i) == int]
print(l4)
l5 = [i for i in l1 if type(i) == str]
print(l5)
```
